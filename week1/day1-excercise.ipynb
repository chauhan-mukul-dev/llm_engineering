```python
import os
from dotenv import load_dotenv
from openai import OpenAI




```

```python
Ollama_URL = "http://localhost:11434/v1"
MODEL = "llama3.2:3b"

openai = OpenAI(base_url=Ollama_URL,api_key="ollama")

message = [{"role":"user" , "content": "can you teach me french!"}]
```

```python
response = openai.chat.completions.create(model=MODEL,messages=message)

response.choices[0].message.content

```

```python
def create_message(user_message):
    messages = [{"role":"system" , "content":"you are a colleage of mine you name is vivek"},
                {"role":"user" , "content":user_message}]
    return messages
```

```python
def chat_wiht_ai(user_message):
    response = openai.chat.completions.create(model=MODEL , messages=create_message(user_message))
    print(response.choices[0].message.content)
```

```python
chat_wiht_ai("how are you")
```

```python
chat_wiht_ai("whts your name")
```
