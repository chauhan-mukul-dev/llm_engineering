To demonstrate your familiarity with OpenAI API, and also Ollama, build a tool that takes a technical question,  
and responds with an explanation. This is a tool that you will be able to use yourself during the course!

```python
# imports
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:3b'
```

```python
# set up environment
OLLAMA_BASE_URL = "http://localhost:11434/v1"
openai = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
```

```python
# here is the question; type over this to ask something new

question = """Please explain what this code does and why:yield from {book.get("author") for book in books if book.get("author")}"""
```

```python
messages = [{"role": "system", "content":"you are a coding expert of python"},
            {"role": "user", "content": question}]

messages
```

```python
# Get gpt-4o-mini to answer, with streaming
```

```python
# Get Llama 3.2 to answer
response = openai.chat.completions.create(model=MODEL_LLAMA, messages=messages)

print(response.choices[0].message.content)
```
